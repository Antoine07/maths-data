```python
import numpy as np
import pandas as pd

url = "https://raw.githubusercontent.com/Antoine07/maths-data/main/docs/data/penguins.csv"
df = pd.read_csv(url)

df.head()
```

```python
# On garde deux espèces et les deux mesures choisies.
# Les lignes incomplètes sur ces colonnes sont retirées.
colonnes = ["bill_length_mm", "flipper_length_mm", "species"]
data = df.loc[
    df["species"].isin(["Adelie", "Gentoo"]),
    colonnes
].dropna().copy()

# On transforme les deux espèces en classes numériques 0 et 1.
data["species"] = data["species"].map({"Adelie": 1, "Gentoo": 0})

# X contient les mesures ; y contient la classe à prédire.
X = data[["bill_length_mm", "flipper_length_mm"]].to_numpy(dtype=float)
y = data["species"].to_numpy(dtype=int)

print("Dimensions de X :", X.shape)
print("Dimensions de y :", y.shape)
print("Classes :", sorted(data["species"].unique()))
data.head()
```
