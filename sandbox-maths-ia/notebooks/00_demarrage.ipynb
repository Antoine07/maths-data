# Démarrage — NumPy, pandas et Matplotlib

Un exemple minimal pour créer un tableau de données et le représenter.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Créer un tableau avec NumPy et pandas

```python
heure = np.arange(24)
consommation = 0.6 + 0.35 * np.sin(2 * np.pi * (heure - 7) / 24) ** 2

donnees = pd.DataFrame({
    "heure": heure,
    "consommation_kwh": consommation,
})
donnees.head()
```

## Tracer les données avec Matplotlib

```python
plt.plot(donnees["heure"], donnees["consommation_kwh"], marker="o")
plt.xlabel("Heure")
plt.ylabel("Consommation (kWh)")
plt.title("Consommation simulée au cours d'une journée")
plt.grid(alpha=0.3)
plt.show()
```
